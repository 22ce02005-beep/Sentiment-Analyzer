```python
import pandas as pd

import nltk

from nltk.sentiment.vader import SentimentIntensityAnalyzer

from nltk.corpus import stopwords

from nltk.tokenize import word_tokenize

from nltk.stem import WordNetLemmatizer
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/pycaret/pycaret/master/datasets/amazon.csv')
```

```python
df.sample(5)
```

```python
pd.set_option('display.max_colwidth', None)
df.sample(5)
```

```python
#(text) bcz the input is text

def preprocess_text(text):

    tokens = word_tokenize(text.lower())
           
    filtered_tokens = [ token for token in tokens if token not in stopwords.words('english')]

    lemmatizer = WordNetLemmatizer()

    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]

    processed_text = ' '.join(lemmatized_tokens)

    return processed_text

df['reviewText'] = df['reviewText'].apply(preprocess_text) 

df.head()
```

```python
#analysing the reviews

analyzer = SentimentIntensityAnalyzer()

def find_sentiment(text):

    score = analyzer.polarity_scores(text) 
    #gives score of negativity, neutrality, positivity & compound

    sentiment = 1 if score['pos'] > 0 else 0
    #checking only for positivity, if anywhere above 0 we're marking it sentiment = 1 , iow, positive

    return sentiment

df['sentiment'] = df['reviewText'].apply(find_sentiment)
df.head()
```

```python
from sklearn.metrics import confusion_matrix, classification_report
print("confusion_matrix: \n")
print(confusion_matrix(df['Positive'], df['sentiment']))
```

```python

print("\n classification_report: \n")
print(classification_report(df['Positive'], df['sentiment'])) 
#96% is a good recall score, 79% accuracy is a good accuracy score too!
```
